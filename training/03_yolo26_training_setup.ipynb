```python

from pathlib import Path
import shutil
import pickle
import joblib

# ReNexa paths on your friend's laptop
PROJECT = Path(r"C:\Users\lalit\ReNexa")

# Trained YOLO26s model from the full training run
SOURCE_MODEL = (
    PROJECT / "training" / "runs" /
    "yolo26s_final" / "weights" / "best.pt"
)

# Destination folder
SAVE_DIR = PROJECT / "models"
SAVE_DIR.mkdir(parents=True, exist_ok=True)

# Check that the trained model exists
if not SOURCE_MODEL.exists():
    raise FileNotFoundError(
        f"Trained model not found: {SOURCE_MODEL}"
    )

# Save a separate copy of the trained model
SAVED_MODEL = SAVE_DIR / "renexa_yolo26s_best.pt"
shutil.copy2(SOURCE_MODEL, SAVED_MODEL)

print("Trained model saved:", SAVED_MODEL)

# Save the current validation metrics
evaluation_metrics = {
    "model": "YOLO26s",
    "split": "validation",
    "precision": 0.7997,
    "recall": 0.6055,
    "f1_score": 0.6892,
    "mAP50": 0.6854,
    "mAP50_95": 0.5366
}

# Save metrics using Joblib
joblib.dump(
    evaluation_metrics,
    SAVE_DIR / "renexa_metrics.joblib"
)

# Save metrics using Pickle
with open(SAVE_DIR / "renexa_metrics.pkl", "wb") as file:
    pickle.dump(evaluation_metrics, file)

print("Joblib metrics saved.")
print("Pickle metrics saved.")
print("All saving operations completed.")
```
